````python
# Provided: analyze_nutrition_main node

def analyze_nutrition_main(state: MainRecipeState) -> MainRecipeState:
    """Analyzes nutrition for the extracted ingredients."""
    ingredients = state["ingredients"]
    
    print("\n" + "="*60)
    print("ANALYZING NUTRITION")
    print("="*60)
    
    # Analyze first 3 ingredients for brevity
    ingredients_to_analyze = ingredients[:3]
    
    system_prompt = """You are a nutritionist. For each ingredient, provide:
    - Estimated calories per serving
    - Key nutrients
    - Any common allergens
    Be concise."""
    
    ingredient_text = "\n".join(
        f"- {ing['quantity']} {ing['name']} ({ing['category']})"
        for ing in ingredients_to_analyze
    )
    
    messages = [
        SystemMessage(content=system_prompt),
        HumanMessage(content=f"Analyze these ingredients:\n{ingredient_text}")
    ]
    
    response = llm.invoke(messages)
    
    return {**state, "nutrition_summary": response.content}

print("analyze_nutrition_main defined!")
````

````python
# Build the main workflow

main_builder = StateGraph(MainRecipeState)
main_builder.add_node("extract_ingredients", extract_ingredients_main)
main_builder.add_node("analyze_nutrition", analyze_nutrition_main)
main_builder.add_edge(START, "extract_ingredients")
main_builder.add_edge("extract_ingredients", "analyze_nutrition")
main_builder.add_edge("analyze_nutrition", END)

recipe_analyzer = main_builder.compile()

print("Main Workflow built!")
print("\nArchitecture:")
print("  extract_ingredients -> analyze_nutrition")
print("\nSubgraph relationship:")
print("  - extract_ingredients invokes: extraction_subgraph")
````
